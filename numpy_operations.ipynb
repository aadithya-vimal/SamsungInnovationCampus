```python
import numpy as np
import pandas as pd

data = np.genfromtxt('Emp_data.csv',
                     delimiter = ',',
                     skip_header = 1,
                     dtype = str)

print(data)
```

```python
#Extract numeric columns
age = data[:,0].astype(float)
salary = data[:,1].astype(float)
experience = data[:,2].astype(float)
performance = data[:,4].astype(float)

numeric_data = pd.DataFrame({
    'Age': age,
    'Salary': salary,
    'Experience': experience,
    'Performance': performance
})

print(numeric_data.head(20))

print(f"age: {age}")
print(f"salary: {salary}")
print(f"experience: {experience}")
print(f"performance: {performance}")
```

```python
#Extracting categorical data
department = data[:,3]
print(f"department: {department}")
```

```python
#Summary statistics
print(f"Mean age: {np.mean(age)}")
print(f"Median salary: {np.median(salary)}")
print(f"Standard deviation of experience: {np.std(experience)}")
print(f"Minimum performance: {np.min(performance)}")
print(f"Maximum performance: {np.max(performance)}")
```

```python
#Filtering salary
high_salary=data[salary>70000]
print(f"High salaries: {high_salary}")
print(f"Number of dimensions: {high_salary.ndim}")
```

```python
#Segmentation
filtered = data[(salary>70000) & (experience>5)]
print(f"Filtered data: {filtered}")
```

```python
#Dept-wise analysis
it_employees = data[department == 'IT']
print(f"IT employees: \n {it_employees}")
```

```python
#Outlier detection
outliers = salary[salary > 90000]
print(f"Salary outliers: \n {outliers}")
```

```python
norm_salary = salary / np.linalg.norm(salary)
print(f"Normalized salaries: \n {norm_salary}")
```

```python
#Coefficient of correlation
corr = np.corrcoef(experience, salary)
print(f"Correlation coefficient: \n {corr}")
```

```python
#New feature creation
salary_per_year = salary / (experience + 1)
print(f"Salary per year: \n {salary_per_year}")
```

```python
#Reshaping for ML algorithm
X = np.column_stack((age, experience))
Y = salary
print(f"X: \n {X}")
print(f"Y: \n {Y}")
```
