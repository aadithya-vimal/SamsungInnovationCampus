```python
def boxplot_iris(feature_names, dataset):
  i=1
  plt.figure(figsize=(11,9))
  for col in feature_names:
    plt.subplot(2,2,i)
    plt.axis('on')
    plt.tick_params(axis='both', left=True, top=False, right= False, bottom= True, labelleft= False, labeltop= False, labelright= False, labelbottom=False)
    dataset[col].plot(kind='box', subplots=True, sharex=False)
    plt.title(col)
    i+=1
  plt.show()
```

```python
boxplot_iris(iris.columns[:-1], iris)
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

data= load_iris()
x=data.data
y= data.target

X_train, X_test, y_train, y_test = train_test_split(x, y, test_size= 0.3, random_state=42)
#finding the optimum point for testing and training
for depth in range(1,11):
  model= DecisionTreeClassifier(criterion='entropy', max_depth=depth)
  model.fit(X_train, y_train)
  train_acc=model.score(X_train, y_train)
  test_acc= model.score(X_test, y_test)
  print(depth, " | ", round(train_acc, 3), " | ", round(test_acc, 3))
```

```python
from sklearn.datasets import load_wine
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

data= load_wine()
x=data.data
y= data.target
print(data.DESCR)
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size= 0.3, random_state=42)
#finding the optimum point for testing and training
print("Depth |  Testing Acc   |  Training Acc")
for depth in range(1,11):
  model= DecisionTreeClassifier(criterion='entropy', max_depth=depth)
  model.fit(X_train, y_train)
  train_acc=model.score(X_train, y_train)
  test_acc= model.score(X_test, y_test)
  print(depth, "    | ", round(test_acc, 3), "        | ", round(train_acc, 3))
```

Math for Linear Regression(single var)

1. find mean of X and Y values
2. find x-mean of x
3. find (x-x mean) (y - y mean)
4. find (x- x mean)^2
5. find sumation of 3. and 4.
6. find beta1 or the slop = sumation of 3./4.
7. find beta0 = intercept= (y mean - beta1 * x mean)

```python
from sklearn.linear_model import LinearRegression
import numpy as np

x= np.array([[1], [2], [3]])
y= np.array([40, 50, 65])
model= LinearRegression()
model.fit(x,y)
print("Intercept: ", model.intercept_)
print("Slop: ", model.coef_[0])
pred= model.predict([[4]])
print(pred)
```
