```python
import numpy as np
arr = np.array([1,3,5,7])
arr2=arr
print(id(arr))
print(id(arr2))
```

```python
#creating an array using a tuple
arr3=np.array((1,2,3,4))
print(arr3)
#creating a array using a dictionary with key value pairs
arr4=np.array({1:'a',2:'b',3:'c'})
print(arr4)
#creating a array using a set
arr5=np.array({1,2,3,4})
print(arr5)
#using range function in arrays
arr6=np.array(range(1,10))
print(arr6)
len10=np.arange(10)   #creates an array in sequential order
len(len10)
len10.size
arr7=np.arange(1,5,0.2)
print(arr7)
id(arr7)
arr7=np.array([[1,2],[2,4],[4,6]])
print(len(arr7))   #len is a function
print(arr7.size)
print(arr7.shape)   #returns in the form of a tuple
print(arr7.ndim)
```

```python
#create a 3D array and check the size and shape of it
arr8=np.array([[[1,2],[3,4],[4,5]],[[2,1],[3,6],[2,8]]])
print(arr8.size)
print(arr8.shape)
```

```python
c= np.array([[1,2],[3,4],[5,6]])
d=np.append(c,[[7,8]],axis=0)
print(d)
new_columns = np.array([[7, 8]] * d.shape[0])
e=np.append(d, new_columns, axis=1)
print(e)
print(np.delete(e,0,axis=0))
np.delete(e,1,axis=1)
```
