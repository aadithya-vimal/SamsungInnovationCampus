# Titanic Dataset Analysis
This notebook provides a detailed exploration of the Titanic dataset using the pandas library.

````python
import pandas as pd

# Load the dataset into a DataFrame
df = pd.read_csv('/content/titanic.csv')
````

### 1. Displaying the first few rows (head)
The `head()` method returns the first n rows for the object based on position. It is useful for quickly testing if your object has the right type of data in it.

````python
display(df.head())
````

### 2. Displaying the last few rows (tail)
The `tail()` method returns the last n rows for the object based on position.

````python
display(df.tail())
````

### 3. Dataset Information
The `info()` method prints information about a DataFrame including the index dtype and columns, non-null values and memory usage.

````python
df.info()
````

### 4. Column Names
Accessing the `columns` attribute returns the column labels of the DataFrame.

````python
print(df.columns)
````

### 5. Index Range
The `index` attribute returns the index (row labels) of the DataFrame.

````python
print(df.index)
````

### 6. Extracting a Single Column
You can extract a single column using bracket notation. Here we extract the 'age' column.

````python
age_column = df['Age']
display(age_column.head())
````

### 7. Extracting Multiple Columns
To extract multiple columns, pass a list of column names inside the brackets.

````python
subset = df[['Age', 'Sex']]
display(subset.head())
````

### 8. Using df.iloc
The `iloc` attribute is used for integer-location based indexing/selection by position. Below we select the first 5 rows and the first 3 columns.

````python
# Selection by position: [rows, columns]
display(df.iloc[0:5, 0:3])
````

### 9. Extracting a Single Value with iloc
To get a specific single value (Series value in a specific row), you can provide single integer coordinates to `iloc`.

````python
# Get the value at row index 0, column index 3 (Name)
single_value = df.iloc[0, 3]
print(f"Value at row 0, column 3: {single_value}")
````

### 10. Renaming Columns with inplace=True
The `rename()` method allows you to change column names. Using `inplace=True` modifies the existing DataFrame directly.

````python
# Rename 'Pclass' to 'Passenger_Class' and 'Survived' to 'Survival_Status'
df.rename(columns={'Pclass': 'Passenger_Class', 'Survived': 'Survival_Status'}, inplace=True)

# Verify the change
print(df.columns)
````

```markdown
### 11. Renaming Columns WITHOUT inplace
By default, `rename()` returns a new DataFrame and does not modify the original one unless `inplace=True` is specified. This is useful for method chaining or preserving the original data.
```

````python
# Rename columns and assign to a new variable
new_df = df.rename(columns={'Sex': 'Gender', 'Fare': 'Ticket_Price'})

print("Original DataFrame columns:")
print(df.columns[4:10])

print("\nNew DataFrame columns:")
print(new_df.columns[4:10])
````
