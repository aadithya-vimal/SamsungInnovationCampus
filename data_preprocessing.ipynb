```python
df2 = df.copy()
df2.dropna(axis=1, thresh=500, inplace=True)
df2.isnull().sum()
```

```python
df3 = df.copy()
df3.isnull().sum()
```

```python
df3.dropna(subset=['age'], axis=0, inplace=True)
display(df2.isnull().sum())
display(df.shape)
```

```python
df4 = df.copy()
df4['age'] = df4['age'].fillna(0)
deck_mode = df4['deck'].mode()[0]
df4['deck'] = df4['deck'].fillna(deck_mode)

# Check the results
print(df4.isnull().sum())
```

```python
df5 = df.copy()
df5['age'] = df5['age'].fillna(df5['age'].median())
deck_mode = df5['deck'].mode()[0]
df5['deck'] = df5['deck'].fillna(deck_mode)

print(df5.isnull().sum())
```

```python
df6 = df.copy()
mode = df6['age'].mode()[0]
df6['age'] = df6['age'].fillna(mode)
deck_mode = df6['deck'].mode()[0]
df6['deck'] = df6['deck'].fillna(deck_mode)
print(df6.isnull().sum())
```

## Outlier handling

```python
from scipy.stats import zscore
df8 = df4.copy()
df8['Z_Score'] = zscore(df8['age'])
outliers = df8.loc[df8['Z_Score']>=3]
print(f"DF with Z Scores: \n {df8}")
print(f"Outliers: \n {outliers}")
```

```python
df9 = df4.copy()
Q1 = df9['age'].quantile(0.25)
Q3 = df9['age'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = df[(df9['age'] < lower_bound) | (df9['age'] > upper_bound)]
print(outliers)
df9_no_outliers = df9[(df9['age'] >= lower_bound) & (df9['age'] <= upper_bound)]
print(df9_no_outliers)
```
