```python
import numpy as np

print("--- Creating Arrays ---")
# Create a 2x3 array of zeros
zeros_array = np.zeros((2, 3))
print("Zeros Array:\n", zeros_array)

# Create a 3x2 array of ones
ones_array = np.ones((3, 2))
print("\nOnes Array:\n", ones_array)

# Create a 2x2 array with random values between 0 and 1
rand_array = np.random.rand(2, 2)
print("\nRandom Array (2x2):\n", rand_array)

# Create a 2x2 array with random values integers
rand_array2 = np.random.randint(1,10,(2, 2))
print("\nRandom Array with random values integers(2x2):\n", rand_array2)

# Create a 2x2 array with random values normal form
rand_array3 = np.random.randn(2, 2)
print("\nRandom Array with random values normal form(2x2):\n", rand_array3)

# Create another 2x2 array for operations
array_a = np.array([[1, 2], [3, 4]])
array_b = np.array([[5, 6], [7, 8]])
print("\nArray A:\n", array_a)
print("\nArray B:\n", array_b)

print("\n--- Matrix Operations ---")
# Transpose
transposed_a = array_a.T
print("\nTranspose of A:\n", transposed_a)

# Dot product using np.dot()
dot_product = np.dot(array_a, array_b)
print("\nDot Product (A . B) using np.dot():\n", dot_product)

# Matrix multiplication using @ operator
matrix_mult = array_a @ array_b
print("\nMatrix Multiplication (A @ B) using @ operator:\n", matrix_mult)
print("\nMatrix Multiplication using np.matmul() operator:\n", np.matmul(array_a, array_b))

# Element-wise Addition
addition = array_a + array_b
print("\nElement-wise Addition (A + B):\n", addition)

# Element-wise Subtraction
subtraction = array_a - array_b
print("\nElement-wise Subtraction (A - B):\n", subtraction)

# Element-wise Multiplication
multiplication = array_a * array_b
print("\nElement-wise Multiplication (A * B):\n", multiplication)
```

```python
import numpy as np

print("Enter elements for Matrix 1 (3x3). Each row should have 3 space-separated numbers.")
matrix1_rows = []
for i in range(3):
    row_str = input(f"Enter row {i+1} for Matrix 1: ")
    matrix1_rows.append(list(map(int, row_str.split())))
matrix1 = np.array(matrix1_rows)
print(f"\nMatrix 1:\n{matrix1}")

print("\nEnter elements for Matrix 2 (3x3). Each row should have 3 space-separated numbers.")
matrix2_rows = []
for i in range(3):
    row_str = input(f"Enter row {i+1} for Matrix 2: ")
    matrix2_rows.append(list(map(int, row_str.split())))
matrix2 = np.array(matrix2_rows)
print(f"\nMatrix 2:\n{matrix2}")

product_matrix = np.dot(matrix1, matrix2)
print(f"\nProduct of Matrix 1 and Matrix 2:\n{product_matrix}")

transposed_product = product_matrix.T
print(f"\nTransposed Product Matrix:\n{transposed_product}")
```

```python
n = int(input("Enter the size of the square matrix (n): "))
matrix = []
print(f"Enter {n} rows for the {n}x{n} matrix, with elements separated by spaces:")
for i in range(n):
    while True:
        row_str = input(f"Enter row {i + 1} (expecting {n} numbers separated by spaces): ")
        row = list(map(int, row_str.split()))
        if len(row) == n:
            matrix.append(row)
            break
        else:
            print(f"Error: You entered {len(row)} numbers. Please enter exactly {n} numbers for this row.")

print("Your square matrix:")
for r in matrix:
    print(r)
```

```python
import numpy as np

n = int(input("Enter the size of the symmetric matrix (n): "))
symmetric_matrix = np.zeros((n, n), dtype=int)

print("Enter elements for the upper triangle (including diagonal) row by row:")
for i in range(n):
    for j in range(i, n):
        value = int(input(f"Enter element M[{i+1}][{j+1}]: "))
        symmetric_matrix[i, j] = value
        symmetric_matrix[j, i] = value # Set the symmetric element

print("\nYour Symmetric Matrix:")
print(symmetric_matrix)
```

# Eigen vectors and Eigen values

### Explanation of Eigenvalues and Eigenvectors

-   **Eigenvalues** are special scalar values associated with a linear system of equations. They are often represented by the Greek letter lambda (λ).

-   **Eigenvectors** are special non-zero vectors that only change by a scalar factor (the eigenvalue) when a linear transformation (like multiplying by a matrix) is applied to them. When you multiply a matrix A by its eigenvector v, you get the same eigenvector back, scaled by the eigenvalue λ (i.e., Av = λv).

-   In the output from the code cell above, the first array (`eigenvalues`) contains the scalar eigenvalues. The second array (`eigenvectors`) contains the corresponding eigenvectors, where each column is an eigenvector linked to the eigenvalue at the same index in the `eigenvalues` array.

```python
import numpy as np

# Define a sample square matrix
# You can change this matrix to any square matrix you like
matrix = np.array([
    [4, 2],
    [1, 3]
])

print("Original Matrix:")
print(matrix)

# Calculate eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(matrix)

print("\nEigenvalues:")
print(eigenvalues)

print("\nEigenvectors (each column corresponds to an eigenvalue):")
print(eigenvectors)
```

# Differentiation


```python
import sympy

print("--- Symbolic Differentiation (using SymPy) ---")

# Define the variables as symbols
x, k = sympy.symbols('x k')

# Function 1: x^2 - 3k
f1_sym = x**2 - 3*k
derivative_f1_sym = sympy.diff(f1_sym, x)
print(f"Original function 1: f(x) = {f1_sym}")
print(f"First derivative of f1 with respect to x: f'(x) = {derivative_f1_sym}")

# Function 2: x^3 - 1
f2_sym = x**3 - 1
derivative_f2_sym = sympy.diff(f2_sym, x)
print(f"\nOriginal function 2: g(x) = {f2_sym}")
print(f"First derivative of f2 with respect to x: g'(x) = {derivative_f2_sym}")

# You can also evaluate the symbolic derivative at a point
point_sym = 2
print(f"\nEvaluating derivative of f1 at x = {point_sym} (assuming k=some_value, here k is treated as a constant with respect to x): {derivative_f1_sym.subs(x, point_sym)}")
print(f"Evaluating derivative of f2 at x = {point_sym}: {derivative_f2_sym.subs(x, point_sym)}")
```

```python
import sympy

# Define the variables as symbols
x, n = sympy.symbols('x n')

# Function 3: (x^3 + 2x)^n
f3_sym = (x**3 + 2*x)**n
derivative_f3_sym = sympy.diff(f3_sym, x)
print(f"\nOriginal function 3: h(x) = {f3_sym}")
print(f"First derivative of h with respect to x: h'(x) = {derivative_f3_sym}")

# Example: evaluate for n=2, x=1
point_x = 1
point_n = 2
evaluated_derivative = derivative_f3_sym.subs({x: point_x, n: point_n})
print(f"\nEvaluating derivative of h at x = {point_x} and n = {point_n}: {evaluated_derivative}")
```

```python
import sympy

x, y = sympy.symbols('x y')

# Define the function
f_xy = sympy.sin(x*y) + sympy.cos(x+y)

print(f"Original function: f(x, y) = {f_xy}")

# Calculate partial derivative with respect to x
derivative_x = sympy.diff(f_xy, x)
print(f"\nPartial derivative with respect to x: ∂f/∂x = {derivative_x}")

# Calculate partial derivative with respect to y
derivative_y = sympy.diff(f_xy, y)
print(f"Partial derivative with respect to y: ∂f/∂y = {derivative_y}")
```
