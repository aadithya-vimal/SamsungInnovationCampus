# Basic use of python libraries


```python
#Using numpy
import numpy as np

arr = np.array([10,20,30,40])
print("Mean of array:",arr.mean())
```

```python
#Using pandas for data handling
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
}

df = pd.DataFrame(data)

print(df)
```

```python
#Using matplotlib for data visualization
import matplotlib.pyplot as plt

marks = [25, 30, 35, 40, 80, 55, 40, 90]

plt.figure(figsize=(3,3))
plt.plot(marks)
plt.xlabel('Subject')
plt.ylabel('Marks')
plt.title('Marks Distribution')
plt.show()
```

```python
#Student result analysis using lamba function
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Kyle', 'Tom', 'Jenny'],
    'Marks': [85, 90, 34, 40, 32, 45, 78]
}

df=pd.DataFrame(data)

df['Result']=df['Marks'].apply(lambda x: 'Pass' if x>=40 else 'Fail')

print(df)
```

```python
#Attendance percentage calculation using
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Kyle', 'Tom', 'Jenny'],
    'Attendance': [85, 90, 34, 40, 32, 45, 78]
}

df = pd.DataFrame(data)

df['Attendance_Percentage']=(df['Attendance']/90)*100

print(df)
```

```python
#Finding class average using groupby function
import pandas as pd

data = {
    'Department': ['EEE', 'DS', 'CSE', 'DS', 'CSE', 'ECE', 'EEE'],
    'Marks': [85, 90, 34, 40, 32, 45, 78]
}

df=pd.DataFrame(data)

class_average=df.groupby('Department')['Marks'].mean()

print(class_average)
```

```python
#Missing data handling(real scenario)
import pandas as pd
import numpy as np

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Kyle', 'Tom', 'Jenny'],
    'Marks': [85, np.nan, 34, np.nan, 32, 45, 78]
}

df=pd.DataFrame(data)

df['Marks'].fillna(df['Marks'].mean(), inplace=True)

print(df)
```

# EDA using python libraries

```python
#Basic EDA
import pandas as pd
import numpy as np

data = {
    'Name': ['Alice', np.nan, 'Charlie', 'David', 'Kyle', 'Tom', 'Jenny'],
    'Marks': [85, 90, np.nan, 40, np.nan, 45, 78]
}

df=pd.DataFrame(data)
print("DESCRIBE")
df.describe().T
print("INFO")
df.info()
print("SHAPE")
df.shape
print("COLUMNS TOLIST")
df.columns.tolist()
print("SUM OF NULLS")
df.isnull().sum()
```

# Mathematical functions in python

```python
from IPython.display import display,Math

symbols = [
    r"\alpha",
    r"\beta",
    r"\gamma",
    r"\sum_{i=1}^{n} x_i",
    r"\int_a^b f(x) dx",
    r"\frac{a}{b}",
    r"x^2 + y^2",
    r"\nable f(x)",
    r"\P(A|B)",
    r"mu",
    r"sigma"
]

for s in symbols:
    display(Math(s))
```

```python
#Vector visualization in python
import matplotlib.pyplot as plt

plt.figure(figsize=(3,3))
plt.quiver(0,0,2,3)
plt.quiver(0,0,4,1)

plt.xlim(0,5)
plt.ylim(0,5)
plt.title("Vector visualization")

plt.grid()
plt.show()
```

```python
#Probability simulation in python
import random

trials = 1000
heads = 0

for _ in range(trials):
  if(random.choice(['H','T'])=='H'):
    heads+=1

print("Probability of heads: ",heads/trials)
```

```python
#Statistics in python
import numpy as np

data = [10,20,30,40,50]

print("Mean:",np.mean(data))
print("Median:",np.median(data))
print("Standard deviation:",np.std(data))
print("Variance:",np.var(data))
```

```python
#Trigonometric functions in python
import math

angle=math.radians(30)
print("Sin(30): ",math.sin(angle))
print("Cos(30): ",math.cos(angle))
print("Tan(30): ",math.tan(angle))
```

```python
#Logarithmic functions in python
import math

x=10

print("log(x): ",math.log(x))
print("log10(x): ",math.log10(x))
```

```python
#Exponential functions in python
import math

x=2

print("e^x= ",math.exp(x))
```

```python
#GCD in python
import math

print("GCD of 5,7= ",math.gcd(5,7))
```

```python
#Plot function in python
import matplotlib.pyplot as plt
import numpy as np

x=np.linspace(-10,10,100)
y=x**2

plt.plot(x,y)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("y=x^2")
plt.grid()
plt.show()
```

# Simple AI Model


```python
from sklearn.linear_model import LinearRegression
import numpy as np

X=np.array([[1],[2],[3],[4]])
Y=np.array([40,50,60,70])

model=LinearRegression()
model.fit(X,Y)

prediction = model.predict([[5]])

print("Predicted marks for the fifth hour: ",prediction)
```
