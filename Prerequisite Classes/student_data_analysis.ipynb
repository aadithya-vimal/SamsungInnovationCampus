```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

data = np.genfromtxt('datasets/Student.csv',
                     delimiter = ',',
                     skip_header = 1,
                     dtype = str)
df = pd.DataFrame(data)
print(df)
print(data)
```

```python
#Extracting columns
StudyHours = data[:,0].astype(float)
print(f"Study Hours: \n{StudyHours}")
Attendance = data[:,1].astype(float)
print(f"Attendance: {Attendance}")
Marks = data[:,2].astype(float)
print(f"Marks: {Marks}")
Sleep = data[:,3].astype(float)
print(f"Sleep: {Sleep}")
```

```python
print(f"Mean marks: {np.mean(Marks)}")
print(f"Median marks: {np.median(Marks)}")
print(f"Standard deviation of marks: {np.std(Marks)}")
```

```python
print(f"Students scoring above average: {np.sum(Marks > np.mean(Marks))} students :-\n {data[Marks>np.mean(Marks)]}")
```

```python
corr1 = np.corrcoef(StudyHours,Marks)
corr2 = np.corrcoef(Attendance,Marks)
print(f"Correlation between Study Hours and Marks: \n {corr1}")
print(f"Correlation between Attendance and Marks: \n {corr2}")
```

```python
normalised_marks = (Marks - np.mean(Marks))/np.std(Marks)
print(f"Normalized marks: \n {normalised_marks}")
```

```python
new_feature = Marks/StudyHours
print(f"Feature(Marks per StudyHours): \n {new_feature}")
```

```python
sorted_marks = np.sort(Marks)
print(f"Sorted marks: \n {sorted_marks}")
print(f"Top Performers: \n", sorted_marks[-10:])
```

```python
X = np.column_stack((StudyHours,Attendance,Sleep))
Y = Marks

model = LinearRegression()
model.fit(X, Y)

a = float(input("Enter Study Hours: "))
b = float(input("Enter Attendance: "))
c = float(input("Enter Sleep: "))
test_data=np.array([[a,b,c]])

print(f"Coefficients: {model.coef_}")
print(f"Intercept: {model.intercept_}")
print('\n')
predictions = model.predict(test_data)
print(f"-----Predictions-----\n Entered Study Hours: {a} \n Entered Attendance : {b} \n Entered Sleep: {c} \nPredicted marks for entered values: {predictions}")
```

```python
plt.bar(StudyHours, Marks)
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.title('Study Hours vs Marks')
plt.show()
```

```python
plt.bar(Attendance, Marks)
plt.xlabel('Attendance')
plt.ylabel('Marks')
plt.title('Attendance vs Marks')
plt.show()
```

```python
plt.bar(Sleep, Marks)
plt.xlabel('Sleep')
plt.ylabel('Marks')
plt.title('Sleep vs Marks')
plt.show()
```
