```python
import pandas as pd
import numpy as np

def load_boston():
    data_url = "http://lib.stat.cmu.edu/datasets/boston"
    raw_df = pd.read_csv(data_url, sep="\\s+", skiprows=22, header=None)
    data = np.hstack([raw_df.values[::2, :], raw_df.values[1::2, :2]])
    target = raw_df.values[1::2, 2]
    return {'data': data, 'target': target}

boston = load_boston()
X = boston['data']
y = boston['target']

# Q7: Solve using Least Squares method: w = (X^T X)^-1 X^T y
# We use np.linalg.lstsq for numerical stability
w, residuals, rank, s = np.linalg.lstsq(X, y, rcond=None)

feature_names = [
    'CRIM', 'ZN', 'INDUS', 'CHAS', 'NOX', 'RM',
    'AGE', 'DIS', 'RAD', 'TAX', 'PTRATIO', 'B', 'LSTAT'
]

print("--- Q7: Least Squares Weight Vector ---")
for name, weight in zip(feature_names, w):
    relation = "proportion" if weight > 0 else "inverse proportion"
    print(f"{name}: {weight:.4f} ({relation})")

print("\n--- Comparison with Q6 ---")
print("In Q6, we used only 4 data points and 4 features, making it a determined system Ax=b.")
print("In Q7, the least-squares method uses all 506 samples, providing a more statistically significant generalization.")
print("Typically, weights in Q7 are smaller in magnitude and more stable than the exact solution in Q6,")
print("which was highly sensitive to the specific 4 samples chosen.")
```

```python
import numpy as np

# Define the rating vectors for users a, b, and c
a = np.array([[4], [5], [2], [2]])
b = np.array([[4], [0], [2], [0]])
c = np.array([[2], [2], [0], [1]])

# Calculate Euclidean distances
dist_ab = np.linalg.norm(a - b)
dist_bc = np.linalg.norm(b - c)
dist_ac = np.linalg.norm(a - c)

print(f"Euclidean distance between a and b: {dist_ab:.4f}")
print(f"Euclidean distance between b and c: {dist_bc:.4f}")
print(f"Euclidean distance between a and c: {dist_ac:.4f}")

distances = {
    ('a', 'b'): dist_ab,
    ('b', 'c'): dist_bc,
    ('a', 'c'): dist_ac
}

closest_pair = min(distances, key=distances.get)
farthest_pair = max(distances, key=distances.get)

print(f"\nThe two users who are the closest are: {closest_pair} (Distance: {distances[closest_pair]:.4f})")
print(f"The two users who are farthest apart are: {farthest_pair} (Distance: {distances[farthest_pair]:.4f})")
```

```python
import numpy as np
from scipy.spatial.distance import cosine

# Define the rating vectors for users a, b, and c
a = np.array([4, 5, 2, 2])
b = np.array([4, 0, 2, 0])
c = np.array([2, 2, 0, 1])

# Calculate Cosine distances
# Cosine distance = 1 - Cosine similarity
dist_ab = cosine(a, b)
dist_bc = cosine(b, c)
dist_ac = cosine(a, c)

print(f"Cosine distance between a and b: {dist_ab:.4f}")
print(f"Cosine distance between b and c: {dist_bc:.4f}")
print(f"Cosine distance between a and c: {dist_ac:.4f}")

distances = {
    ('a', 'b'): dist_ab,
    ('b', 'c'): dist_bc,
    ('a', 'c'): dist_ac
}

closest_pair = min(distances, key=distances.get)
farthest_pair = max(distances, key=distances.get)

print(f"\nThe two users who are the closest (smallest cosine distance) are: {closest_pair} (Distance: {distances[closest_pair]:.4f})")
print(f"The two users who are farthest apart (largest cosine distance) are: {farthest_pair} (Distance: {distances[farthest_pair]:.4f})")
```

```python
import numpy as np

# Define the matrix D
D = np.array([[2, 1],
              [1, 2]])

# The characteristic equation is det(D - lambda*I) = 0
# For a 2x2 matrix [[a, b], [c, d]], this is:
# (a-lambda)(d-lambda) - bc = 0
# lambda^2 - (a+d)lambda + (ad-bc) = 0

# Calculate eigenvalues using NumPy's linalg.eigvals
eigenvalues = np.linalg.eigvals(D)

print("Matrix D:")
print(D)
print(f"\nThe eigenvalues calculated from the characteristic equation are: {eigenvalues}")

# Verification:
# Trace(D) = 2+2 = 4. Sum of eigenvalues = 3+1 = 4.
# Det(D) = 2*2 - 1*1 = 3. Product of eigenvalues = 3*1 = 3.
```

```python
import numpy as np

def logistic_inverse(y):
    """
    Computes the logit function, which is the inverse of the logistic function.
    x = ln(y / (1 - y))
    """
    return np.log(y / (1 - y))

# Example calculation
y_val = 0.5
x_inv = logistic_inverse(y_val)

print(f"The inverse of the logistic function (logit) for y = {y_val} is x = {x_inv}")
print(f"Formula: x = ln(y / (1 - y))")
```

```python
import sympy
from sympy import symbols, diff, exp, log

# --- Part 1 (Question 12a) ---
print("--- Part 1(Question 12a): First Derivatives ---")
x, k, a, b = symbols('x k a b')

# Function 1: f(x) = x³ - 1
f1 = x**3 - 1
df1 = diff(f1, x)
print(f"1) f(x) = x³ - 1  => f'(x) = {df1}")

# Function 2: f(x) = log(log(x² - 3k))
f2 = log(log(x**2 - 3*k))
df2 = diff(f2, x)
print(f"2) f(x) = log(log(x² - 3k)) => f'(x) = {df2}")

# Function 3: f(x) = exp(ax^b)
f3 = exp(a * x**b)
df3 = diff(f3, x)
print(f"3) f(x) = exp(ax^b) => f'(x) = {df3}")

# --- Part 2 (Question 12b) ---
print("\n--- Part 2(Question 12b): Partial Derivatives ---")
y = symbols('y')
f_xy = exp(x**2 + 2*y**2)

# First-order
fx = diff(f_xy, x)
fy = diff(f_xy, y)

# Second-order
fxx = diff(fx, x)
fxy = diff(fx, y)
fyx = diff(fy, x)
fyy = diff(fy, y)

print(f"f(x, y) = exp(x² + 2y²)")
print(f"fx  = {fx}")
print(f"fy  = {fy}")
print(f"fxx = {fxx}")
print(f"fxy = {fxy}")
print(f"fyx = {fyx}")
print(f"fyy = {fyy}")
```

```python
import sympy
from sympy import symbols, diff, exp, log

# Define SymPy symbols
x, y, k, a, b = symbols('x y k a b')

print("--- Problem 13: Derivatives f'(x) ---")

# A. f(x) = x^3 - 1
f_a = x**3 - 1
df_a = diff(f_a, x)
print(f"A. f(x) = x^3 - 1  => f'(x) = {df_a}")

# B. f(x) = log(log(x^2 - 3k))
f_b = log(log(x**2 - 3*k))
df_b = diff(f_b, x)
print(f"B. f(x) = log(log(x^2 - 3k)) => f'(x) = {df_b}")

# C. f(x) = exp(ax^b)
f_c = exp(a * x**b)
df_c = diff(f_c, x)
print(f"C. f(x) = exp(ax^b) => f'(x) = {df_c}")

print("\n--- Problem 14: Partial Derivatives for f(x, y) = exp(x^2 + 2y^2) ---")

f_xy = exp(x**2 + 2*y**2)

# First-order partial derivatives
fx = diff(f_xy, x)
fy = diff(f_xy, y)

# Second-order partial derivatives
fxx = diff(fx, x)
fxy = diff(fx, y)
fyx = diff(fy, x)
fyy = diff(fy, y)

print(f"fx  = {fx}")
print(f"fy  = {fy}")
print(f"fxx = {fxx}")
print(f"fxy = {fxy}")
print(f"fyx = {fyx}")
print(f"fyy = {fyy}")
```
