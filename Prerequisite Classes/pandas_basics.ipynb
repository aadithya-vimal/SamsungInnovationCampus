# Operations in Pandas

```python
exam_data = {
    'Physics' : [67,98,72,45],
    'Chemistry' : [67,98,72,45],
    'Math' : [67,98,72,45],
    'Bio' : [67,98,72,45]
}
print(f"Created 'exam_data' dictionary: \n {exam_data}")
```

```python
#Creating dataframe
import pandas as pd
df = pd.DataFrame(exam_data, index = ['S1', 'S2', 'S3', 'S4'])
print(f"\n Created 'df' dataframe: \n\n {df}")
print("\nTable format:\n")
display(df)
```

```python
#Deleting/dropping a row
df2 = df3 = df.copy()
df2.drop('S1', inplace=True)
df3.drop('Math', inplace=True, axis=1)
print(f"Original dataframe: \n {df}")
print(f"\nNew dataframe with deleted row: \n {df2}")
print(f"\nNew dataframe with deleted column: \n {df3}")
```

```python
#Accessing values of single row
df4 = df.copy()
display(df.loc['S2'])
display(df.iloc[2])
```

```python
#Extracting filtered rows
df5 = df.copy()
df6 = df.copy()
display(df5.iloc[[0,2]])
display(df6.loc[: , 'Math'])
```

```python
#Renaming index
df7 = df.copy()
df7.index = ['Asha', 'Bala', 'Charan', 'Darsh']
display(df7)
```

```python
#Accessing real time data
df7.iloc[0, :3]
```

```python
df8 = df7.copy()
# Create a new row with the sum of each column
df8.loc['Total'] = df8.sum()

display(df8)
```

```python
df9 = df.copy()
df9.reset_index(inplace=True)
display(df9)
df9.rename(columns={'index' : 'Names'}, inplace = True)
display(df9)
```

```python
df10 = df7.copy()
df10.loc['Eevee'] = {
    'Physics' : 67,
    'Chemistry' : 98,
    'Math' : 72,
    'Bio' : 45
}
df10.loc['Total'] = df10.sum()
df10.reset_index(inplace=True)
df10.rename(columns={'index' : 'Names'}, inplace = True)
display(df10)
```

```python
df11 = df7.copy()
df11.loc['Asha','Physics'] =100
display(df11)
```

```python
df12=df7.copy()
t_data = df12.T
display(t_data)
display(df12)
```

```python
df13=df10.copy()
sorted_data_p=df13.sort_values(by='Physics', ascending=False)
sorted_data_n=df13.sort_values(by='Names', ascending=False)
print(f"Sorted data: (Physics)\n\n {sorted_data_p}")
print(f"\nSorted data: (Names)\n\n {sorted_data_n}")
```

```python
df14=df10.copy()
# Select the row where 'Names' is 'Asha'
asha_data = df14[df14['Names'] == 'Asha']

# Select the numeric columns from asha_data
numeric_cols = ['Physics', 'Chemistry', 'Math', 'Bio']
asha_numeric_df = asha_data[numeric_cols]

sum_val = asha_numeric_df.sum(axis=1).iloc[0]
prod_val = asha_numeric_df.prod(axis=1).iloc[0]
diff_val = (asha_numeric_df['Physics'] - asha_numeric_df['Chemistry'] - asha_numeric_df['Math'] - asha_numeric_df['Bio']).iloc[0]
fract = asha_numeric_df / 100

# Print the results directly
print(f"Sum: {sum_val}")
print(f"Product: {prod_val}")
print(f"Difference: {diff_val}")
print(f"Fraction:\n{fract.to_string(index=False)}")
```

```python
#Merging data
a_values = {'Name':['A', 'B', 'C', 'D'], 'Gender': ['M', 'F', 'M', 'F'], 'Age': [20,24,21,24]}
b_values = {'Name':['E', 'A', 'G', 'B'], 'Position': ['Assistant', 'Leader', 'Sweeper', 'Security'], 'Salary':[564654,98782,2132132,89896]}
df_a = pd.DataFrame(a_values)
df_b = pd.DataFrame(b_values)
pd.merge(df_a,df_b,on='Name',how='outer')
```

```python
#Concatenating data
pd.concat([df_a,df_b],axis=1)
```

```python
tdf4=tdf.copy()
tdf4_analyzed=pd.pivot_table(tdf4, index=['who'], columns='survived', values='age', aggfunc='count')
tdf4_analyzed['Total']=tdf4_analyzed.sum(axis=1)
tdf4_analyzed.loc['Total']=tdf4_analyzed.sum(axis=0)
display(tdf4_analyzed)
```
