# Vectors for Machine Learning

Vectors are the fundamental way we represent data in Machine Learning. In this guide, we'll learn how to build and manipulate them from scratch.

### 1. What is a Vector?
In programming, a vector is simply an **ordered list of numbers**. In Machine Learning:
- A **Row Vector** is a list written horizontally: `[1, 2, 3]`
- A **Column Vector** is a list written vertically.

We use the `numpy` library because it is optimized for these mathematical structures.

### 1a. Creating a Row Vector
Think of a row vector as a single data point in a 1D space.

```python
import numpy as np
row_vec = np.array([1, 2, 3])
print("Row Vector:", row_vec)
```

### 1b. Creating a Column Vector
In NumPy, we use double brackets `[[ ]]` to represent rows and columns for a vertical vector.

```python
col_vec = np.array([[4], [5], [6]])
print("Column Vector:\n", col_vec)
```

### 1c. Vector Addition
When we add vectors of the same size, NumPy adds them element-by-element.

```python
a = np.array([1, 2])
b = np.array([10, 20])
print("Addition [1,2] + [10,20] =", a + b)
```

### 1d. Scalar Multiplication: Scaling a Vector
Multiplying a vector by a scalar (a single number) stretches or shrinks the vector by multiplying every element by that number.

```python
import numpy as np
# Define a vector
v = np.array([1, 2, 3])
# Multiply by a scalar (3)
scaled_v = v * 3
print("Original Vector:", v)
print("Scaled Vector (v * 3):", scaled_v)
```

```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.cluster import KMeans
from sklearn.neighbors import KNeighborsClassifier
from sklearn.feature_extraction.text import CountVectorizer

# --- 1. Creating Vectors From Scratch ---

# 1a. Creating a simple Row Vector (1D Array)
# Think of this as a single point in space or one record.
vector_row = np.array([1, 2, 3])

# 1b. Creating a Column Vector (2D Array)
# Notice the extra brackets [[ ]]. This tells NumPy it has 3 rows and 1 column.
vector_col = np.array([[4],
                       [5],
                       [6]])

# 1c. Basic Arithmetic (Element-wise)
# When we add vectors, we add the numbers in the same positions.
# [1, 2] + [10, 20] = [11, 22]
a = np.array([1, 2])
b = np.array([10, 20])
vec_sum = a + b

# 1d. Scalar Multiplication
# Multiplying a vector by a single number (a 'scalar') stretches or shrinks it.
# 3 * [1, 2] = [3, 6]
scaled = a * 3

print("Row Vector:", vector_row)
print("Column Vector:\n", vector_col)
print("Sum of [1,2] + [10,20]:", vec_sum)
print("Scaled [1,2] by 3:", scaled)
```

```python
import numpy as np
# --- 2a. Dot Product ---
v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])

# Dot Product: (1*4) + (2*5) + (3*6) = 32
dot_product = np.dot(v1, v2)
print("Dot Product:", dot_product)
```

```python
# --- 2b. L2 Norm (Magnitude) ---
# The 'length' of a vector
magnitude = np.linalg.norm(v1)
print("Magnitude of v1:", magnitude)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
# --- 3a. Cosine Similarity ---
# Reshape is needed because sklearn expects a 2D array (matrix)
v1_2d = v1.reshape(1, -1)
v2_2d = v2.reshape(1, -1)

sim = cosine_similarity(v1_2d, v2_2d)
print("Cosine Similarity:", sim[0][0])
```

```python
# --- 3b. Euclidean Distance ---
# The straight-line distance between two points
dist = np.linalg.norm(v1 - v2)
print("Euclidean Distance:", dist)
```

```python
from sklearn.cluster import KMeans
# --- 4. K-Means Clustering ---
X = np.array([[1, 2], [1, 4], [10, 2], [10, 4]])
kmeans = KMeans(n_clusters=2, n_init='auto').fit(X)
print("Cluster Labels:", kmeans.labels_)
```

## Real-World Applications of Vectors

Vectors aren't just math; they power the modern world:

1.  **Recommendation Systems**: Netflix represents movies as vectors (genre, length, actors). It uses **Cosine Similarity** to find movies 'pointing' in the same direction as your favorites.
2.  **Feature Engineering (PCA)**: Large datasets (1000s of variables) are hard to process. PCA squashes long vectors into shorter ones while keeping the important info.
3.  **Game Dev (Physics)**: Characters move using vectors. Velocity is a vector (speed + direction). To move a character, you simply add a `movement_vector` to their `position_vector`.
4.  **Time Series Analysis**: Stock prices over a week can be a 7-element vector. Machine learning looks for patterns in these vectors to predict the next day.
