## 1. Quick theory notes

A **continuous random variable** can take infinitely many values in an interval.

### PDF
The **probability density function** is:
$$f(x)$$

For a continuous variable, the probability at exactly one point is:
$$P(X=x)=0$$

### CDF
The cumulative distribution function is:
$$F(x)=P(X\le x)$$

Probabilities over intervals are computed using the CDF:
$$P(a \le X \le b) = F(b)-F(a)$$


## 2. Uniform distribution

A uniform distribution assumes every value in an interval is equally likely.

### Real-world examples
- waiting time if a bus arrives at a random time within a fixed interval
- random initialization within a fixed range
- random sampling over an interval


```python
# Uniform distribution U(10,20)
a = 10
b = 20
scale = b - a

x = np.linspace(8, 22, 500)
pdf_vals = st.uniform.pdf(x, loc=a, scale=scale)
cdf_vals = st.uniform.cdf(x, loc=a, scale=scale)

print("PDF at x=15:", st.uniform.pdf(15, loc=a, scale=scale))
print("P(X<=14):", st.uniform.cdf(14, loc=a, scale=scale))
print("P(12<=X<=18):", st.uniform.cdf(18, loc=a, scale=scale) - st.uniform.cdf(12, loc=a, scale=scale))
print("Median:", st.uniform.ppf(0.5, loc=a, scale=scale))
```

```python
# Plot PDF and CDF for uniform distribution
plt.figure(figsize=(8,4))
plt.plot(x, pdf_vals)
plt.xlabel("x")
plt.ylabel("Density")
plt.title("Uniform PDF: U(10,20)")
plt.grid(True, alpha=0.3)
plt.show()

plt.figure(figsize=(8,4))
plt.plot(x, cdf_vals)
plt.xlabel("x")
plt.ylabel("P(X≤x)")
plt.title("Uniform CDF: U(10,20)")
plt.grid(True, alpha=0.3)
plt.show()
```

## 3. Normal distribution

The normal distribution is one of the most important continuous models.

### Real-world examples
- measurement errors
- exam scores
- sensor noise
- model latency after aggregation


```python
# Normal distribution N(mu, sigma^2)
mu = 50
sigma = 8

x = np.linspace(20, 80, 500)
pdf_vals = st.norm.pdf(x, loc=mu, scale=sigma)
cdf_vals = st.norm.cdf(x, loc=mu, scale=sigma)

print("PDF at x=50:", st.norm.pdf(50, loc=mu, scale=sigma))
print("P(X<=55):", st.norm.cdf(55, loc=mu, scale=sigma))
print("P(45<=X<=60):", st.norm.cdf(60, loc=mu, scale=sigma) - st.norm.cdf(45, loc=mu, scale=sigma))
print("5th percentile:", st.norm.ppf(0.05, loc=mu, scale=sigma))
print("95th percentile:", st.norm.ppf(0.95, loc=mu, scale=sigma))
```

```python
# Plot PDF and CDF for normal distribution
plt.figure(figsize=(8,4))
plt.plot(x, pdf_vals)
plt.xlabel("x")
plt.ylabel("Density")
plt.title("Normal PDF: N(50, 8²)")
plt.grid(True, alpha=0.3)
plt.show()

plt.figure(figsize=(8,4))
plt.plot(x, cdf_vals)
plt.xlabel("x")
plt.ylabel("P(X≤x)")
plt.title("Normal CDF: N(50, 8²)")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Standardization and z-score
x_value = 62
z = (x_value - mu) / sigma
prob = st.norm.cdf(z)

print("z-score:", z)
print("P(X <= 62):", prob)
```

### Real-world problem
Suppose exam scores in a machine learning class are approximately normal with mean 68 and standard deviation 10.
Students scoring above 85 are considered outstanding.
This can be modeled using the normal distribution and solved using the CDF or z-scores.


## 4. Exponential distribution

The exponential distribution is commonly used for **waiting time** between random events.

### Real-world examples
- time until the next customer arrives
- time until a server fails
- waiting time between support calls


```python
# Exponential distribution with rate lambda = 0.5
rate = 0.5
scale = 1 / rate

x = np.linspace(0, 15, 500)
pdf_vals = st.expon.pdf(x, scale=scale)
cdf_vals = st.expon.cdf(x, scale=scale)

print("PDF at x=2:", st.expon.pdf(2, scale=scale))
print("P(X<=3):", st.expon.cdf(3, scale=scale))
print("P(2<=X<=5):", st.expon.cdf(5, scale=scale) - st.expon.cdf(2, scale=scale))
print("Median:", st.expon.ppf(0.5, scale=scale))
```

```python
# Plot PDF and CDF for exponential distribution
plt.figure(figsize=(8,4))
plt.plot(x, pdf_vals)
plt.xlabel("x")
plt.ylabel("Density")
plt.title("Exponential PDF with rate 0.5")
plt.grid(True, alpha=0.3)
plt.show()

plt.figure(figsize=(8,4))
plt.plot(x, cdf_vals)
plt.xlabel("x")
plt.ylabel("P(X≤x)")
plt.title("Exponential CDF with rate 0.5")
plt.grid(True, alpha=0.3)
plt.show()
```

### Real-world problem
A chatbot server receives a request on average every 4 seconds.
If inter-arrival times are modeled exponentially, we can compute the probability that the next request arrives within 2 seconds or after 10 seconds.


## 5. Chi-square distribution

The chi-square distribution appears often in:
- variance estimation
- goodness-of-fit tests
- independence tests


```python
# Chi-square distribution with df = 6
df = 6
x = np.linspace(0, 25, 500)
pdf_vals = st.chi2.pdf(x, df=df)
cdf_vals = st.chi2.cdf(x, df=df)

print("PDF at x=5:", st.chi2.pdf(5, df=df))
print("P(X<=8):", st.chi2.cdf(8, df=df))
print("P(4<=X<=10):", st.chi2.cdf(10, df=df) - st.chi2.cdf(4, df=df))
print("95th percentile:", st.chi2.ppf(0.95, df=df))
```

```python
# Plot PDF and CDF for chi-square distribution
plt.figure(figsize=(8,4))
plt.plot(x, pdf_vals)
plt.xlabel("x")
plt.ylabel("Density")
plt.title("Chi-square PDF with df=6")
plt.grid(True, alpha=0.3)
plt.show()

plt.figure(figsize=(8,4))
plt.plot(x, cdf_vals)
plt.xlabel("x")
plt.ylabel("P(X≤x)")
plt.title("Chi-square CDF with df=6")
plt.grid(True, alpha=0.3)
plt.show()
```

## 6. t-distribution

The t-distribution is important when the sample size is small and the population variance is unknown.

### Real-world use
- confidence intervals for small samples
- one-sample and two-sample t-tests


```python
# Compare t-distributions for different degrees of freedom
x = np.linspace(-5, 5, 500)

plt.figure(figsize=(8,4))
for df in [2, 5, 10, 30]:
    plt.plot(x, st.t.pdf(x, df=df), label=f"df={df}")
plt.plot(x, st.norm.pdf(x), label="Normal", linestyle='--')
plt.xlabel("x")
plt.ylabel("Density")
plt.title("t-distribution vs Normal")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 7. F-distribution

The F-distribution is used when comparing variances and in ANOVA.

### Real-world use
- comparing variability of two algorithms
- testing differences across multiple group means


```python
# F-distribution example
d1, d2 = 5, 10
x = np.linspace(0.01, 5, 500)

plt.figure(figsize=(8,4))
plt.plot(x, st.f.pdf(x, dfn=d1, dfd=d2))
plt.xlabel("x")
plt.ylabel("Density")
plt.title(f"F-distribution PDF with d1={d1}, d2={d2}")
plt.grid(True, alpha=0.3)
plt.show()

print("P(X<=2):", st.f.cdf(2, dfn=d1, dfd=d2))
print("95th percentile:", st.f.ppf(0.95, dfn=d1, dfd=d2))
```

## 8. Choosing the right continuous distribution

### Uniform
Use when all values in an interval are equally likely.

### Normal
Use when data are symmetric and concentrated around a mean.

### Exponential
Use for waiting time between random events.

### Chi-square
Use for variance-related problems and certain hypothesis tests.

### t-distribution
Use for sample-mean inference when variance is unknown and sample size is small.

### F-distribution
Use when comparing variances or multiple group means.
