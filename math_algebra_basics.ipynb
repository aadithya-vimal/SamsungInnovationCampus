# Algebraic function in python


First Degree Equations

```python
#Solve (x-2)^3
import sympy as sp
x=sp.symbols('x')

exp=(x-2)**3

expanded_expr=sp.expand(exp)

print("Expanded expression: ",expanded_expr)
```

```python
#Solve 2x+5=15
import sympy as sp
x=sp.symbols('x')

exp=2*x+5-15

print("X value:", sp.solve(exp,x))
```

Second Degree Equations

```python
#Solve 2x^2+4x+5=10
import sympy as sp
x=sp.symbols('x')

exp=(2*x)**2+4*x+5-10

print("X value:", sp.solve(exp,x))
```

```python
#Solve x^2-5x=(-6)
import sympy as sp
x=sp.symbols('x')

exp=(x)**2-5*x+6

print("X value:", sp.solve(exp,x))
```

# Sequence


```python
#Arithmetic sequence program

#Take input from user
a=float(input("Enter the first term: "))
d=float(input("Enter the common difference: "))
n=int(input("Enter the number of terms: "))

sequence =[]

for i in range(n):
  term=a+i*d
  sequence.append(term)

print("The sequence is: ",sequence)

sum = (n/2)*(2*a+(n-1)*d)

print("The sum of the sequence is: ",sum)
```

```python
#Geometric sequence program

#Take input from user
a = float(input("Enter the first term (a): "))
r = float(input("Enter the common ratio (r): "))
n = int(input("Enter the number of terms (n): "))

sequence = []
current_term = a
for i in range(n):
  sequence.append(current_term)
  current_term *= r

print("The sequence is:", sequence)

if r == 1:
  sum_sequence = n * a
else:
  sum_sequence = a * (1 - r**n) / (1 - r)

print("The sum of the sequence is:", sum_sequence)
```

```python
elements_str = input("Enter elements for the set, separated by commas: ")

# Split the input string by commas
elements = elements_str.split(',')

# Create a set from the list of elements
user_set = set(elements)

print("The set is:", user_set)

# Check if an element exists in the set
check_element = input("Enter an element to check for in the set: ")

if check_element in user_set:
  print(f"'{check_element}' exists in the set.")
else:
  print(f"'{check_element}' does not exist in the set.")
```

```python
# Get input for the first set
elements_str1 = input("Enter elements for the first set, separated by commas: ")
elements1 = elements_str1.split(',')
set1 = set(elements1)
print("First Set:", set1)

# Get input for the second set
elements_str2 = input("Enter elements for the second set, separated by commas: ")
elements2 = elements_str2.split(',')
set2 = set(elements2)
print("Second Set:", set2)

# Perform set operations
union_set = set1.union(set2)
intersection_set = set1 | set2
difference_set1_2 = set1 & set2 # Elements in set1 but not in set2
difference_set2_1 = set2 - set1 # Elements in set2 but not in set1

# Display results
print("\nUnion of the sets:", union_set)
print("Intersection of the sets:", intersection_set)
print("Difference (Set1 - Set2):", difference_set1_2)
print("Difference (Set2 - Set1):", difference_set2_1)
```

# Concept of functions

```python
import numpy as np
import matplotlib.pyplot as plt
x=np.linspace(-5,5,1001)
fx=3*x+1
print("Domain: All real numbers(-infinity to +infinity)")
print("Range: All real numbers(-infinity to +infinity)")
plt.plot(x, fx, label= "f(x)=3x+1")
plt.axhline(0)
plt.axvline(0)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Graph of f(x)=3x+1")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Function transformation
import numpy as np
import matplotlib.pyplot as plt
x=np.linspace(-5,5,400)
f=x
shift_right = x-1
shift_left = x+1
shift_up = x+2
shift_down = x-2
plt.plot(x,f,label="f(x)=x")
plt.plot(x,shift_right,label="f(x)=x-1")
plt.plot(x,shift_left,label="f(x)=x+1")
plt.plot(x,shift_up,label="f(x)=x+2")
plt.plot(x,shift_down,label="f(x)=x-2")
plt.axhline(0)
plt.axvline(0)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Function Transformation")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Composite functions
import numpy as np
import matplotlib.pyplot as plt
def f(x):
  return x**2+10
def g(x):
  return 3+x
def fog(x):
  return f(g(x))
def gof(x):
  return g(f(x))

np.linspace(-5,5,1000)

plt.plot(x,fog(x),label="f(g(x))")
plt.plot(x,gof(x),label="g(f(x))")
plt.axhline(0)
plt.axvline(0)

plt.xlabel("Y")
plt.ylabel("x")
```

```python
#Inverse function of logisitic regression
import sympy as sp
x,y=sp.symbols('x,y')
expr=1/(1+sp.exp(-x))
inverse=sp.solve(sp.Eq(expr,y))
print("Inverse function: ", inverse)
```
